```python
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

```python
df = pd.concat([train, test])
```

```python
df.head(2)
```

```python
df = pd.get_dummies(df, columns=['person_home_ownership', 'loan_intent', 'loan_grade'], dtype='int64', drop_first=True)
```

```python
df['loan_status'].fillna(df['loan_status'].mean(), inplace=True)
```

```python
df.head(2)
```

```python
sns.heatmap(df.isnull())
```

```python
df = df.drop(['cb_person_default_on_file'], axis=1)
```

# Scale and train data

```python
from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()
scaler.fit(df)
df1 = scaler.transform(df)
```

```python
train_len = train.shape[0]
X_train = df1[:train_len]
X_test = df1[train_len:]
y_train = train['loan_status']
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
```

# cross validation

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import make_scorer, r2_score
```

```python
def test_model(model, X_train, y_train):
    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    r2 = make_scorer(r2_score)
    r2_val_score = cross_val_score(model, X_train, y_train, cv=cv, scoring=r2)
    score = [r2_val_score.mean()]
    return score
```

# Linear regration

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
test_model(lr, X_train, y_train)
```

```python
from sklearn.linear_model import Ridge, Lasso
ridge = Ridge(alpha=0.1)
test_model(ridge, X_train, y_train)
```

```python
lasso = Lasso(alpha=0.1)
test_model(lasso, X_train, y_train)
```

# Support Vector Machine Regration

```python
from sklearn.svm import SVR
svr = SVR()
test_model(svr, X_train, y_train)
```

# Dissition Tree Regration

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor(random_state=42)
test_model(dt, X_train, y_train)
```

# Random Forest Regration

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=42, n_estimators=1000)
test_model(rf, X_train, y_train)
```

# Baggin and Busting

```python
from sklearn.ensemble import BaggingRegressor, GradientBoostingRegressor
br = BaggingRegressor(random_state=42, n_estimators=1000)
gb = GradientBoostingRegressor(random_state=42, n_estimators=1000)
```

```python
test_model(br, X_train, y_train)
```

```python
test_model(gb, X_train, y_train)
```

# fit best model

```python
lr.fit(X_train, y_train)
```
